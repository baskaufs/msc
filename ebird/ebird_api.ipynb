```python
# -----------------------------------------
# Import modules.
# -----------------------------------------

import requests
from typing import List, Dict, Tuple, Optional, Any
from pathlib import Path
import json
from time import sleep

# -----------------------------------------
# Global variables.
# -----------------------------------------

BASE_URL = 'https://api.ebird.org/v2/'
API_ACCESS_KEY_FILENAME = 'ebird_api_key.txt'
```

```python
# Read the API access key from a file in the home directory. The file should contain only the key and no other text.
home = str(Path.home()) # gets path to home directory; supposed to work for both Win and Mac
with open(home + '/' + API_ACCESS_KEY_FILENAME, 'r') as file:
    api_access_key = file.read().strip() # remove any leading or trailing white space or newlines

#print(api_access_key)

HTTP_HEADER = {
    'x-ebirdapitoken': api_access_key
    }

```

Run either the next cell or the one that follows depending on which information you want. Then run the request cell to get the response.

```python
# Recent nearby observations

endpoint = 'data/obs/geo/recent'
request_url = BASE_URL + endpoint
print(request_url)

request_params = {}

latitude = '40.89993'
longitude = '-83.90836'
request_params['lat'] = latitude
request_params['lng'] = longitude

days_back = '1'
request_params['back'] = days_back

distance = '5' # radius in km
request_params['dist'] = distance

print(request_params)
```

```python
# Recent nearby observations of a species

endpoint = 'data/obs/geo/recent/'
species_code = 'boboli'
request_url = BASE_URL + endpoint + species_code
print(request_url)

latitude = '40.89993'
longitude = '-83.90836'
request_params['lat'] = latitude
request_params['lng'] = longitude
days_back = '1'
request_params['back'] = days_back
distance = '50' # radius in km
request_params['dist'] = distance
print(request_params)
```

After generating the URL and parameters using one of the previous cells, run the following cell to get the data.

```python
r = requests.get(request_url, params=request_params, headers=HTTP_HEADER)
print(r.status_code)
#print(r.text)
print(json.dumps(r.json(), indent = 2))
```
